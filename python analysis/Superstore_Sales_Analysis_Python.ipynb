Now starting with different types of analysis

```python
#creating a year column for better understanding
df["Year"] = df["Order Date"].dt.year #creating the column
df[["Order Date", "Year"]].head() #checking the column
```

```python
df["Month_Number"] = df["Order Date"].dt.month #creating a month number column
df["Month"] = df["Order Date"].dt.month_name() #creating a month name column
df[["Order Date", "Year", "Month_Number", "Month"]].head() #checking the results
```

```python
df["Quarter"] = df["Order Date"].dt.quarter
df[["Order Date", "Quarter"]].head()
#creating a quarter column and checking it
```

```python
#calculating the yearly sales
yearly_sales = df.groupby("Year")["Sales"].sum().sort_index()
yearly_sales
```

```python
#Calculating YEARLY PROFIT
yearly_profit = df.groupby("Year")["Profit"].sum().sort_index()
yearly_profit
```

```python
#calculating YEARLY QUANTITY
yearly_quantity = df.groupby("Year")["Quantity"].sum().sort_index()
yearly_quantity
```

```python
#displaying all the 3 fields together
yearly_analysis = df.groupby("Year").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum"),
    Quantity=("Quantity", "sum")
)

yearly_analysis
```

```python
#Calculating the YEARLY PROFIT MARGIN
yearly_analysis["Profit Margin"] = (
    yearly_analysis["Profit"] / yearly_analysis["Sales"]
)

#DISPLAYING IT
yearly_analysis
```

```python
#Calculate Year-over-Year sales growth
yearly_analysis["Sales Growth %"] = (
    yearly_analysis["Sales"].pct_change() * 100
)
yearly_analysis
```

```python
#Find the highest-sales year
highest_sales_year = yearly_analysis["Sales"].idxmax()
highest_sales_value = yearly_analysis["Sales"].max()

print("Highest Sales Year:", highest_sales_year)
print("Sales:", highest_sales_value)

#Find the highest-profit year
highest_profit_year = yearly_analysis["Profit"].idxmax()
highest_profit_value = yearly_analysis["Profit"].max()

print("Highest Profit Year:", highest_profit_year)
print("Profit:", highest_profit_value)

#Find the highest-growth year
highest_growth_year = yearly_analysis["Sales Growth %"].idxmax()
highest_growth = yearly_analysis["Sales Growth %"].max()

print("Highest Sales Growth Year:", highest_growth_year)
print("Growth:", highest_growth, "%")
```

```python
#Create monthly sales analysis
monthly_sales = df.groupby(
    ["Month_Number", "Month"]
)["Sales"].sum().reset_index()

monthly_sales = monthly_sales.sort_values("Month_Number")

monthly_sales

#Create monthly profit analysis
monthly_profit = df.groupby(
    ["Month_Number", "Month"]
)["Profit"].sum().reset_index()

monthly_profit = monthly_profit.sort_values("Month_Number")

monthly_profit

#Combine monthly sales and profit
monthly_analysis = df.groupby(
    ["Month_Number", "Month"]
).agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum")
).reset_index()

monthly_analysis = monthly_analysis.sort_values("Month_Number")

monthly_analysis
```

```python
#Calculate monthly profit margin
monthly_analysis["Profit Margin"] = (
    monthly_analysis["Profit"] / monthly_analysis["Sales"]
)
monthly_analysis

#Find the highest-sales month
highest_sales_month = monthly_analysis.loc[
    monthly_analysis["Sales"].idxmax()
]

print("Highest Sales Month:", highest_sales_month["Month"])
print("Sales:", highest_sales_month["Sales"])
```

```python
#Create a yearly sales chart
plt.figure(figsize=(8, 5))

plt.plot(
    yearly_analysis.index,
    yearly_analysis["Sales"],
    marker="o"
)

plt.title("Yearly Sales Trend")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.grid(True)

plt.show()
```

```python
#Create a monthly sales chart
plt.figure(figsize=(10, 5))

plt.plot(
    monthly_analysis["Month"],
    monthly_analysis["Sales"],
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
category_analysis_final = df.groupby("Category").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum"),
    Quantity=("Quantity", "sum"),
    Order_Count=("Order ID", "count")
)

category_analysis_final["Profit Margin"] = (
    category_analysis_final["Profit"] /
    category_analysis_final["Sales"]
)

category_analysis_final["Sales Contribution %"] = (
    category_analysis_final["Sales"] /
    df["Sales"].sum()
) * 100

category_analysis_final
```

```python
regional_analysis_final = df.groupby("Region").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum"),
    Quantity=("Quantity", "sum"),
    Order_Count=("Order ID", "count")
)

regional_analysis_final["Profit Margin"] = (
    regional_analysis_final["Profit"] /
    regional_analysis_final["Sales"]
)

regional_analysis_final["Sales Contribution %"] = (
    regional_analysis_final["Sales"] /
    df["Sales"].sum()
) * 100

regional_analysis_final
```

```python
customer_analysis_final = df.groupby(
    ["Customer ID", "Customer Name"]
).agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum"),
    Quantity=("Quantity", "sum"),
    Order_Count=("Order ID", "count")
)

customer_analysis_final["Profit Margin"] = (
    customer_analysis_final["Profit"] /
    customer_analysis_final["Sales"]
)

customer_analysis_final
```
